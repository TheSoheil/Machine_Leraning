# Q3-Soheil_Hekmat

0. Dependencies

```python
import numpy as np
import matplotlib.pyplot as plt
```

1. Defining the functions (classes)

```python
def g1(x, y):
    return 2*x - y + 1

def g2(x, y):
    return -x + 2*y - 2

def g3(x, y):
    return -x - y + 3
```

2. Creating a grid to visualize the boundries

```python
x = np.linspace(-5, 5, 500)
y = np.linspace(-5, 5, 500)
X, Y = np.meshgrid(x, y)

G1 = g1(X, Y)
G2 = g2(X, Y)
G3 = g3(X, Y)
```

3. Classifying based on maximum discriminant value

```python
scores = np.stack([G1, G2, G3], axis=2)
labels = np.argmax(scores, axis=2)  # 0→C1, 1→C2, 2→C3
```

4. Plotting the boundaries & regions

```python
plt.figure(figsize=(8, 8))

# decision regions
plt.contourf(X, Y, labels, levels=[-0.5,0.5,1.5,2.5], alpha=0.4,
             colors=['lightblue', 'lightgreen', 'lightcoral'])

# decision boundaries (where gi = gj)
plt.contour(X, Y, G1 - G2, levels=[0], colors='blue', linewidths=2)
plt.contour(X, Y, G1 - G3, levels=[0], colors='green', linewidths=2)
plt.contour(X, Y, G2 - G3, levels=[0], colors='red', linewidths=2)
```

5. Plotting Random points for each class

```python
np.random.seed(0)

# Class 1 random points
pts1 = np.random.uniform(-5, 5, size=(200, 2))
lbls1 = np.argmax(np.stack([
    g1(pts1[:,0], pts1[:,1]),
    g2(pts1[:,0], pts1[:,1]),
    g3(pts1[:,0], pts1[:,1])
], axis=1), axis=1)
pts1 = pts1[lbls1 == 0]
plt.scatter(pts1[:,0], pts1[:,1], c='blue', s=10, label="Class 1")

# Class 2 random points
pts2 = np.random.uniform(-5, 5, size=(200, 2))
lbls2 = np.argmax(np.stack([
    g1(pts2[:,0], pts2[:,1]),
    g2(pts2[:,0], pts2[:,1]),
    g3(pts2[:,0], pts2[:,1])
], axis=1), axis=1)
pts2 = pts2[lbls2 == 1]
plt.scatter(pts2[:,0], pts2[:,1], c='green', s=10, label="Class 2")

# Class 3 random points
pts3 = np.random.uniform(-5, 5, size=(200, 2))
lbls3 = np.argmax(np.stack([
    g1(pts3[:,0], pts3[:,1]),
    g2(pts3[:,0], pts3[:,1]),
    g3(pts3[:,0], pts3[:,1])
], axis=1), axis=1)
pts3 = pts3[lbls3 == 2]
plt.scatter(pts3[:,0], pts3[:,1], c='red', s=10, label="Class 3")
```

6. Plotting everything together (final result)

```python
plt.figure(figsize=(8, 8))

# decision regions
plt.contourf(X, Y, labels, levels=[-0.5,0.5,1.5,2.5], alpha=0.4,
             colors=['lightblue', 'lightgreen', 'lightcoral'])

# decision boundaries (where gi = gj)
plt.contour(X, Y, G1 - G2, levels=[0], colors='blue', linewidths=2)
plt.contour(X, Y, G1 - G3, levels=[0], colors='green', linewidths=2)
plt.contour(X, Y, G2 - G3, levels=[0], colors='red', linewidths=2)


np.random.seed(0)

# Class 1 random points
pts1 = np.random.uniform(-5, 5, size=(200, 2))
lbls1 = np.argmax(np.stack([
    g1(pts1[:,0], pts1[:,1]),
    g2(pts1[:,0], pts1[:,1]),
    g3(pts1[:,0], pts1[:,1])
], axis=1), axis=1)
pts1 = pts1[lbls1 == 0]
plt.scatter(pts1[:,0], pts1[:,1], c='blue', s=10, label="Class 1")

# Class 2 random points
pts2 = np.random.uniform(-5, 5, size=(200, 2))
lbls2 = np.argmax(np.stack([
    g1(pts2[:,0], pts2[:,1]),
    g2(pts2[:,0], pts2[:,1]),
    g3(pts2[:,0], pts2[:,1])
], axis=1), axis=1)
pts2 = pts2[lbls2 == 1]
plt.scatter(pts2[:,0], pts2[:,1], c='green', s=10, label="Class 2")

# Class 3 random points
pts3 = np.random.uniform(-5, 5, size=(200, 2))
lbls3 = np.argmax(np.stack([
    g1(pts3[:,0], pts3[:,1]),
    g2(pts3[:,0], pts3[:,1]),
    g3(pts3[:,0], pts3[:,1])
], axis=1), axis=1)
pts3 = pts3[lbls3 == 2]
plt.scatter(pts3[:,0], pts3[:,1], c='red', s=10, label="Class 3")

plt.title("Decision Boundaries and Decision Regions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
